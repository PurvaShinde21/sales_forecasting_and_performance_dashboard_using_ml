# 02 - Exploratory Data Analysis
## Superstore Sales Dataset


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style='whitegrid')
```

```python
df = pd.read_csv('../dataset/superstore_cleaned.csv')
df['Order Date'] = pd.to_datetime(df['Order Date'])
print(f'Shape: {df.shape}')
```

## Sales by Category


```python
cat_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
cat_sales.plot(kind='bar', color=['#3498db','#e74c3c','#2ecc71'], figsize=(8,5))
plt.title('Sales by Category')
plt.ylabel('Sales ($)')
plt.tight_layout()
plt.show()
```

## Monthly Sales Trend


```python
monthly = df.groupby(df['Order Date'].dt.to_period('M'))['Sales'].sum().reset_index()
monthly['Order Date'] = monthly['Order Date'].dt.to_timestamp()
plt.figure(figsize=(12,5))
plt.plot(monthly['Order Date'], monthly['Sales'], color='#e74c3c', linewidth=2)
plt.fill_between(monthly['Order Date'], monthly['Sales'], alpha=0.15, color='#e74c3c')
plt.title('Monthly Sales Trend')
plt.ylabel('Sales ($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Sales by Region


```python
reg_sales = df.groupby('Region')['Sales'].sum()
plt.figure(figsize=(8,5))
reg_sales.plot(kind='pie', autopct='%1.1f%%', colors=['#1abc9c','#3498db','#9b59b6','#e67e22'])
plt.title('Sales by Region')
plt.ylabel('')
plt.tight_layout()
plt.show()
```

## Discount vs Profit


```python
plt.figure(figsize=(8,5))
plt.scatter(df['Discount'], df['Profit'], alpha=0.3, c=df['Sales'], cmap='RdYlGn', s=20)
plt.colorbar(label='Sales')
plt.xlabel('Discount')
plt.ylabel('Profit')
plt.title('Discount vs Profit')
plt.axhline(y=0, color='red', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

## Top 10 Sub-Categories by Sales


```python
sub_sales = df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=True).tail(10)
plt.figure(figsize=(10,6))
sub_sales.plot(kind='barh', color=sns.color_palette('viridis', 10))
plt.title('Top 10 Sub-Categories by Sales')
plt.xlabel('Sales ($)')
plt.tight_layout()
plt.show()
```
